```python
import joblib
model=joblib.load('Ai_and_Real_Face_Recognize.sav')
```

```python

import gradio as gr
import numpy as np
import cv2

lable_dict = {
    0: 'Real Images',
    1: 'AI-Generated Images',
}

def process_image(image):
    if image is None:
        return "Please upload an image first!"

    # Convert uploaded image to NumPy array
    img_array = np.array(image)

    # Convert RGB image to grayscale
    img_array = cv2.cvtColor(img_array, cv2.COLOR_RGB2GRAY)

    # Resize image to 8x8 pixels
    img_array = cv2.resize(img_array, (8, 8))

    # Reshape for model prediction
    img_array = np.reshape(img_array, (1, 64))

    # Predict character
    result = model.predict(img_array)[0]
    label = lable_dict[result]

    return f"Predicted Character: {label}"


# Build Gradio interface
with gr.Blocks() as demo:

    gr.Markdown("## Sinhala Character Recognition")

    # Upload image instead of drawing on canvas
    image_input = gr.Image(
        label="Upload Character Image",
        type="numpy"
    )

    submit_btn = gr.Button("Predict")

    output_text = gr.Textbox(label="Prediction Result")

    submit_btn.click(
        fn=process_image,
        inputs=image_input,
        outputs=output_text
    )

# Launch application
if __name__ == "__main__":
    demo.launch()
```
