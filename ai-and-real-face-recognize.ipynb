```python
import os
import numpy as np

data_path='/kaggle/input/datasets/kaustubhdhote/human-faces-dataset/Human Faces Dataset'

categories = os.listdir(data_path)
labels=[i for i in range(len(categories))]

category_dict={'Real Images':0,'AI-Generated Images':1}

print(categories)
print(labels)
print(category_dict)
```

```python
import cv2

data=[]
target=[]

for category in categories:
    imgs_path=os.path.join(data_path,category)
    img_names=os.listdir(imgs_path)

    for img_name in img_names:
        img_path=os.path.join(imgs_path,img_name)
        img=cv2.imread(img_path,0)
        img=cv2.resize(img,(8,8))
        data.append(img)
        target.append(category_dict[category])
```

```python
data=np.array(data)
print(data.shape)
```

```python
target=np.array(target)
print(target.shape)
```

```python
data=data.reshape(data.shape[0],data.shape[1]*data.shape[2])
print(data.shape)
```

```python
np.save('data',data)
np.save('target',target)
```

```python
import numpy as np

dat=np.load('data.npy')
target=np.load('target.npy')
```

```python
from sklearn.model_selection import train_test_split
```

```python
train_data,test_data,train_target,test_target=train_test_split(data,target,test_size=0.3)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model=KNeighborsClassifier()
```

```python
model.fit(train_data,train_target)
```

```python
predicted_target=model.predict(test_data)
```

```python
from sklearn.metrics import accuracy_score
```

```python
acc=accuracy_score(test_target,predicted_target)

print("accuracy : ",acc)
```
